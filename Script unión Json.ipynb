### Ahora preparamos el código para que cargue cualquier número de archivos


```python
import glob ## me lee del disco todos los archivos que cumplen ese patron 
import os
```

```python
CARPETA =  r'/Users/ana/Documents/GitHub/Python-PowerBI-Esenciales/datasets' ## si estuviesemos trabajando en power BI sería poner un C:
PATRO_FITXERS = '*_nai_bcn.json'
```

```python
fitxers = sorted (
    glob.glob(os.path.join(CARPETA, PATRO_FITXERS))
)

if not fitxers:
    raise FileNotFoundError(
        f"No hay fichero en {CARPETA}"
    )
```

```python
llista = [pd.read_json(f) for f in fitxers]
dades = pd.concat(llista, ignore_index=True)
```
