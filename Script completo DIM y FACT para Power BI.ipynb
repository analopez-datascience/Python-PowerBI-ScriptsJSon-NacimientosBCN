```python
import glob
import os
import pandas as pd

CARPETA = r"/Users/ana/Documents/GitHub/Python-PowerBI-Esenciales/datasets"
PATRO_FITXERS = "*_nai_bcn.json"

fitxers = sorted(
    glob.glob(os.path.join(CARPETA, PATRO_FITXERS))
)

if not fitxers:
    raise FileNotFoundError(
        f"No hi ha fitxers JSON a {CARPETA}"
    )

#Unir json
llista = [pd.read_json(f) for f in fitxers]
dades = pd.concat(llista, ignore_index=True)

# Dimensió districtes
districtes = (
    dades[["Codi_Districte","Nom_Districte"]]
    .drop_duplicates()
    .sort_values("Codi_Districte")
    .reset_index(drop=True)
)

barris = (
    dades[["Codi_Barri", "Nom_Barri", "Codi_Districte"]]
    .drop_duplicates()
    .sort_values("Codi_Barri")
    .reset_index(drop=True)
)
dades['Valor'] = pd.to_numeric(dades['Valor'], errors='coerce')
naixements = dades.drop(columns=["AEB","Seccio_Censal","Nom_Districte","Nom_Barri"])
naixements = (
    naixements.groupby(
        ["Any", "Codi_Districte","Codi_Barri"], as_index=False
    )["Valor"].sum()
)

print(districtes.head())
print(barris.head())
print(naixements.head())
```
